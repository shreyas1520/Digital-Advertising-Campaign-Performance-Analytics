# 02 - Data Cleaning & Preprocessing

## Objective

Validate data quality, verify business rules, prepare the dataset for exploratory data analysis, and create additional features if required.

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv("C:/Users/Hp/OneDrive/Desktop/Digital Advertising Campaign Performance Analytics/Data/tech_advertising_campaigns_dataset.csv")

df.head()
```

```python
df[['start_date']]
```

```python
df['start_date'] = pd.to_datetime(df['start_date'])
```

```python
df.info()
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns

(df[numeric_cols] < 0).sum()
```

```python
(df['clicks'] > df['impressions']).sum()
```

```python
(df['conversions'] > df['clicks']).sum()
```

```python
(df['revenue'] < 0).sum()
```

```python
(df['ad_spend'] < 0).sum()
```

```python
(df['profit'] < 0).sum()
```

```python
Q1 = df['ad_spend'].quantile(0.25)
Q3 = df['ad_spend'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['ad_spend'] < lower) | (df['ad_spend'] > upper)]

print(outliers.shape)
```

```python
df['Year'] = df['start_date'].dt.year
```

```python
df['Month'] = df['start_date'].dt.month_name()
```

```python
df['Quarter'] = df['start_date'].dt.quarter
```

```python
df['Weekday'] = df['start_date'].dt.day_name()
```

```python
df.to_csv("C:/Users/Hp/OneDrive/Desktop/Digital Advertising Campaign Performance Analytics/Data/tech_advertising_campaigns_dataset.csv", index=False)
```
